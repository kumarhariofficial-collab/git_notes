## Set Functions

1.add 2.clear 3.copy 4.difference 5.difference_update 6.discard 7.intersection 8.intersection_update 9.isdisjoint 10.issubset 11.issuperset 12.pop 13.remove 14.symmetric_difference 15.symmetric_difference_update 16.union 17.update

```python
name = {"hari","kalai","gokul","sathish"}
name.add("rajeev")
name
```

```python
name = {"hari","kalai","gokul","sathish"}
name.clear()
name
```

```python
names = {"hari","kalai","gokul","sathish","amudhan"}
updated_names = names.copy()
updated_names
```

```python
names = {"hari","kalai","gokul","sathish"}
updated_names = {"hari","kalai","gokul","sathish","amudhan","rajeev"}
print(updated_names.difference(names))
```

```python
names = {"hari","kalai","gokul"}
updated_names = {"hari","kalai","gokul","sathish","amudhan","rajeev"}
updated_names.difference_update(names)
print(updated_names)
```

```python
updated_names = {"hari","kalai","gokul","sathish","amudhan","rajeev"}
updated_names.discard("hari")
print(updated_names)
```

```python
names = {"hari","kalai","gokul"}
updated_names = {"hari","kalai","gokul","sathish","amudhan","rajeev"}
result = updated_names.intersection(names)
result
```

```python
names = {"hari","kalai","gokul"}
updated_names = {"hari","kalai","gokul","sathish","amudhan","rajeev"}
names.intersection(updated_names)
print(names)
```

```python
sciene_marks = {100,98,70,60,50}
maths_marks = {99,70,50,89,95}
maths_marks.intersection_update(sciene_marks)
maths_marks
```

```python
sciene_marks = {100,98,70,60,50}
maths_marks = {99,70,50,89,95}
print(maths_marks.isdisjoint(sciene_marks))
```

```python
sciene_marks = {100,98,70,60,50}
maths_marks = {99,75,51,89,95}
print(maths_marks.isdisjoint(sciene_marks))
```

```python
sciene_marks = {100,98,70,60,50}
maths_marks = {99,75,51,89,95}
print(maths_marks.issubset(sciene_marks))
```

```python
sciene_marks = {100,98,70,60,50,99,75,51,89,95}
maths_marks = {99,75,51,89,95}
print(maths_marks.issubset(sciene_marks))
```

```python
sciene_marks = {100,98,70,60,50,99,75,51,89,95}
maths_marks = {99,75,51,89,95}
print(sciene_marks.issuperset(maths_marks))
```

```python
sciene_marks = {100,98,70,60,50,99,75,51,89,95}
maths_marks = {99,75,51,89,95}
print(maths_marks.issuperset(sciene_marks))
```

```python
maths_marks = {99,75,51,89,95}
maths_marks.pop()
print(maths_marks)
```

```python
sciene_marks = {100,98,70,60,50}
sciene_marks.remove(100)
print(sciene_marks)
```

```python
sciene_marks = {100,98,70,60,50}
maths_marks = {99,70,50,89,95,35}
print(sciene_marks.symmetric_difference(maths_marks))
```

```python
sciene_marks = {100,98,70,60,50}
maths_marks = {99,70,50,89,95,80,100,45}
maths_marks.symmetric_difference_update(sciene_marks)
print(maths_marks)
```

```python
science_students = {"hari", "gokul", "kalai"}
math_students = {"kalai", "sathish", "rajeev"}
all_students = science_students.union(math_students)
print(all_students)
```

```python
science_students = {"hari", "gokul", "kalai"}
math_students = {"kalai", "sathish", "rajeev"}
science_students.update(math_students)
print(science_students)
```

```python
student = {"name": "Hari", "age": 22}
result = student.setdefault("course", "Python")
print(student)
```

```python
data = [("Hari", "Python"), ("Mukesh", "Java"), ("Dharani", "Python")]
result = {}
for name, course in data:
    result.setdefault(course, []).append(name)
print(result)
```
